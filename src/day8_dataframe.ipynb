```python
#datafream创建方式
import pandas as pd
import numpy as np

#通过series创建
s1 = pd.Series([1,3,5,7,9])
s2 = pd.Series([2,4,6,8,10])
df = pd.DataFrame({"第一列":s1,"第二列":s2})
type(df)

#通过字典创建
data = {"第一列":[1,2,3,4,5],"第二列":[10,20,30,40,50]}
df = pd.DataFrame(data)
type(df)

#通过列表创建
data = [[1,10],[2,20],[3,30],[4,40],[5,50]]
df = pd.DataFrame(data,columns=["第一列","第二列"])
type(df)

#通过numpy数组创建
data = np.array([[1,10],[2,20],[3,30],[4,40],[5,50]])
df = pd.DataFrame(data,columns=["第一列","第二列"])
type(df)



```
