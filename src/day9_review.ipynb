```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'seg_id': ['s1', 's2', 's3', 's4', 's5', 's6', 's6'],
'weather':['晴天', '雨天', '夜间', '雨天', '晴天', '夜间', '夜间'],
'avg_speed':[58.5, 32.0, np.nan, 24.5, 65.0, 18.0, 18.0],
'temp_note':['ok', 'ok', 'err', 'ok', 'ok', 'ok', 'ok'],
'takeover_cnt': [0, 2, 1, 3, 0, 4, 4]})

df = df.drop(columns='temp_note')
df = df.rename(columns={'avg_speed': 'speed_kmh'})
df = df.drop_duplicates()
df = df[~df['speed_kmh'].isna()]
high_risk = df[(df['weather'].isin(['雨天', '夜间'])) & (df['takeover_cnt'] >= 2)]
high_risk = high_risk.sort_values(by=['takeover_cnt','speed_kmh'], ascending= [False,True])
high_risk = high_risk.reset_index(drop=True)
new_df = high_risk.nlargest(2, 'takeover_cnt')[['seg_id', 'takeover_cnt']]
# new_df
df.describe()
```
