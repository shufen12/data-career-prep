```python
import pandas as pd
import numpy as np
#Series 属性
s = pd.Series([1, 3, 5, 7, 9], index=['a', 'b', 'c', 'd', 'e'])
print(s) #打印Series
print(s.index) #获取索引
print(s.values) #获取值
print(s.name) #获取名称
print(s.shape) #获取形状
print(s.ndim) #获取维度
print(s.size) #获取元素个数
print(s.dtype) #获取数据类型
print(s.loc['a']) #获取索引为'a'的元素,显式索引，可以切片
print(s.iloc[0:3:1]) #获取第一个元素，隐式索引，可以切片
print(s.at['a']) #获取索引为'a'的元素,显式索引,不可以切片
print(s.iat[0]) #获取第一个元素，隐式索引，不可以切片
```

```python
#访问数据
print(s['a']) #通过索引访问数据
print(s[s<5]) #通过条件访问数据
print(s.head(3)) #获取前3个元素,默认打印前5行 s.head()
print(s.tail(3)) #获取后3个元素，默认打印后5行 s.tail()
```

```python
#常用方法
print(s.describe()) #获取描述性统计信息
print(s.sum(axis=0)) #计算总和,axis=0表示按列求和，axis=1表示按行求和,默认axis=0
print(s.cumsum()) #计算累计和
print(s.cummax()) #计算累计最大值
print(s.cummin()) #计算累计最小值
print(s.cumprod()) #计算累计乘积
print(s.mean()) #计算平均值
print(s.std()) #计算标准差
print(s.var()) #计算方差
print(s.min()) #计算最小值
print(s.max()) #计算最大值
print(s.pct_change()) #计算收益率 （当日收盘价/前日收盘价 - 1）
print(s.mode()) #获取众数
print(s.median()) #获取中位数
print(s.quantile(0.5)) #获取分位数
print(s.diff()) #获取两个元素之间的差值
print(s.abs()) #获取绝对值
print(s.nlargest(3, columns=['a','b'])) #表示取最大的三个数，在dataframe中，s.nlargest(3, columns=['a','b'])，表示取列a中最大的三个数，然后再按照列b排序
print(s.nsmallest(3)) #表示取最小的三个数
print(s.sample(3)) #随机获取三个元素
print(s.resample()) #重新采样分组
print(s.replace(1, 100)) #替换值
print(s.sort_values(ascending=True)) #按值排序，如果ascending=True，则按升序排序，如果ascending=False，则按降序排序
print(s.sort_index()) #按索引排序，在dataframe中，s.sort_index(by=['a', 'b'])，表示先按照列索引a排序，再按照索引b排序
print(s.keys()) #获取索引
print(s.value_counts()) #获取值的计数
print(s.duplicated()) #判断是否有重复值，True表示有，False表示没有
print(s.drop_duplicates()) #删除重复值
print(s.unique()) #获取唯一值
print(s.nunique()) #获取有几个唯一值，就是去重后的元素个数
print(s.isin([1, 3])) #判断在Series中是否有指定值，True表示有，False表示没有
print(s.isna()) #判断是否为缺失值，False表示不是缺失值，True表示是缺失值 如果写成 ~s.isna()，则表示取反，即判断是否不是缺失值
```

```python
range_km = pd.Series([520, 610, 480, None, 610, 390],index=['秦PLUS', 'Model3', 'AionS', '测试车X', '极氪001', '海豚'])
print(range_km.size) #获取元素个数
print(range_km.isna().sum()) #获取缺失值的个数
print(range_km.mean()) #计算平均值
print(range_km.max()) #计算最大值
print(range_km[(500<=range_km) & (range_km<=600)]) #获取指定范围的元素
print()
print(range_km.sort_values(ascending=False)) #按值排序,从高到低
print()
print(range_km.value_counts()) #获取值的计数
```

```python
import numpy as np
np.random.seed(42)
scores = pd.Series(np.random.randint(50,101,10),index=['学生'+str(i) for i in range(1,11)])
avg = scores.mean()
print("平均分：",avg)
print("最高分：",scores.max())
print("最低分：",scores.min())
print("高于平均分人数：",scores[avg < scores].count())
```

```python
#温度案例分析
temperatures = pd.Series([28,31,29,32,30,27,33],index=["周一","周二","周三","周四","周五","周六","周日"])
print(temperatures.mean()) #计算平均值
print(temperatures.sort_values(ascending=False)) #将温度从高到低排序
t_diff = temperatures.diff().abs() #计算Series变化值的绝对值
t_diff.sort_values(ascending=False) #把差值从大到小排序
t_diff.sort_values(ascending=False).index[:2] #拿到前两个变化值的索引
t_diff.sort_values(ascending=False).index[:2].tolist() #将series对象转化为列表对象
print("温度变化最大的两天：",*(t_diff.sort_values(ascending=False).index[:2].tolist())) #将列表对象解包

```

```python
#股票案例分析
prices = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2023-01-01',periods=10)) #periods 表示次数
a = prices.pct_change() #计算收益率
print(a.idxmax()) #获取最大收益率对应的索引
print(a.idxmin()) #获取最低收益率对应的索引
print(a.std()) #计算波动率（收益率的标准差）
```

```python
#销售数据分析
import pandas as pd
sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=pd.date_range('2026-01-01',periods=12,freq='MS')) #freq 表示频率，这里是按月，默认是天
sales.resample('QS').mean() #以每一个季度开始重新分组并计算平均值
print('销量最高的月份：',sales.idxmax()) #找到销量最高的月份
a = sales.pct_change() #月环比的增长率
b = a>0 #获取布尔值
b.rolling(3) #滑动窗口，这里表示滑动窗口大小为3
print(b[b.rolling(3).sum() == 3].index.tolist()) #找到连续增长三个月的月份 
```

```python
#每小时销售额数据分析
import numpy as np
import pandas as pd
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0,100,24),index=pd.date_range('2026-01-01',periods=24,freq='h'))
day_sales = hourly_sales.resample('D').sum() #按照每天重采样计算一天的总销售额
business_hourly_sales = hourly_sales.between_time('8:00','22:00').sum() #计算从八点到晚上十点的销售额 between_time('time','time')
print(business_hourly_sales / (day_sales - business_hourly_sales)) #计算营业时间的销售额与非营业时间销售额的比例
print(hourly_sales.nlargest(3)) #表示取最大的三个数
```
